Diferente de listas, tuplas são **imutáveis** - não dá pra modificar depois de criadas. Útil pra dados que não devem mudar.

*Tuplas:*

```python
quadrados = [n**2 for n in range(10)]
pares = [n for n in range(20) if n % 2 == 0]
quadrados, pares
```
