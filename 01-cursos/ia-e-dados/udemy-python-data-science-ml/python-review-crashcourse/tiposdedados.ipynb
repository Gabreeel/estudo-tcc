```python
funcoes_suspeitas_a = {'CreateProcess', 'WriteFile', 'RegSetValue', 'CreateThread'}
funcoes_suspeitas_b = {'CreateThread', 'VirtualAlloc', 'WriteFile', 'GetProcAddress'}
print(f'União: {funcoes_suspeitas_a | funcoes_suspeitas_b}')
print(f'Interseção: {funcoes_suspeitas_a & funcoes_suspeitas_b}')
print(f'Só em A: {funcoes_suspeitas_a - funcoes_suspeitas_b}')
```

Desempacotamento é útil quando funções retornam múltiplos valores tipo (resultado, confiança, tempo):

```python
hashes_truncados = [hash(i) % 1000 for i in range(10)]
portas_suspeitas = [p for p in range(1024, 65536) if p % 8080 == 0]
hashes_truncados, portas_suspeitas[:5]
```
