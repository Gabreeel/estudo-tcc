## NumPy Array Indexing and Selection

```python
import numpy as np
```

```python
arrI = np.arange(0,11)
```

```python
arrI
```

```python
arrI[8]
```

```python
print(arrI[1:5]) #Slicing, returns elements from index 1 to 4
print(arrI[0:5]) #Slicing, returns elements from index 0 to 4
print(arrI[:6]) #Slicing, returns elements from index 0 to 5
print(arrI[5:]) #Slicing, returns elements from index 5 to the end
```

```python
print(arrI)

slice_of_arrI = arrI[0:5]
slice_of_arrI[:] = 100

print(slice_of_arrI)
print(arrI)#O slice_of_arrI é uma view do arrI, então quando alteramos o slice_of_arrI, o arrI também é alterado
```

```python
arrI_copy = arrI.copy() #Diferente do slice, o copy cria uma cópia do array, então quando alteramos o copy, o arrI não é alterado
```

```python
arrI_copy[:] = 200
```

```python
print(arrI)
print(arrI_copy)
```

```python
arr_2d = np.array([[5,10,15],[20,25,30],[35,40,45]])
```

```python
arr_2d
```

```python
print(arr_2d[1][1]) #Retorna o elemento da segunda linha e segunda coluna, que é 25
print(arr_2d[0]) #Retorna a primeira linha do array, que é [5,10,15]
print(arr_2d[0:2]) #Retorna as duas primeiras linhas do array, que são [[5,10,15],[20,25,30]]
print(arr_2d[2][1]) #Retorna o elemento da terceira linha e segunda coluna, que é 40

# Outra notação para acessar elementos de arrays 2D é usando vírgula para separar os índices de linha e coluna:

print(arr_2d[1,1]) #Retorna o elemento da segunda linha e segunda coluna, que é 25
print(arr_2d[:2,1:]) #Retorna as duas primeiras linhas e todas as colunas a partir da segunda coluna, que são [[10,15],[25,30]]
print(arr_2d[1:,:2]) #Retorna todas as linhas a partir da segunda linha e as duas primeiras colunas, que são [[20,25],[35,40]]
```

## NumPy Operations

    - Array with Array
    - Array with Scalars
    - Universal Array Functions

```python
import numpy as np
```

```python
arr = np.arange(0,11)
```

```python
arr + arr
```

```python
arr - arr
```

```python
arr * arr
```

```python
print(arr + 100)
print(arr * 2)
print(arr ** 2) #Eleva cada elemento do array ao quadrado
```

```python
arr / arr #retorna nan para o elemento 0, pois não é possível dividir por 0
```

```python
np.sqrt(arr) #Retorna a raiz quadrada de cada elemento do array
```

```python
np.exp(arr) #Retorna o exponencial de cada elemento do array
```

```python
np.log(arr) #Retorna o logaritmo natural de cada elemento do array
```

```python
np.sin(arr) #Retorna o seno de cada elemento do array
```
