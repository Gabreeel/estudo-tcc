## NumPy Array Indexing and Selection

Aqui começa uma parte muito importante: **acessar pedaços específicos do array**.

Isso parece básico, mas é o tipo de coisa que fica aparecendo toda hora em análise de dados. Em malware, por exemplo, posso querer pegar só um intervalo de bytes, só uma região de uma matriz, só as primeiras linhas de uma imagem, ou só os valores que passam em uma condição.

Essa parte é sobre pegar exatamente o pedaço que eu quero, sem precisar fazer gambiarra com loop manual o tempo inteiro.

```python
import numpy as np
```

**Array simples para treinar indexação**

Agora o notebook cria um array de 0 até 10. Esse é o tipo de exemplo bom para aprender slicing sem se perder.

```python
arrI = np.arange(0,11)
```

```python
arrI
```

**Acessando por índice**

Assim como listas, arrays começam no índice 0.

Então `arrI[8]` pega o elemento que está na posição 8.

```python
arrI[8]
```

**Slicing**

Slicing é pegar um pedaço do array.

A regra principal é: o início entra, o fim não entra. Então `[1:5]` pega os índices 1, 2, 3 e 4.

```python
print(arrI[1:5]) # slicing: retorna os elementos do índice 1 até o 4
print(arrI[0:5]) # slicing: retorna os elementos do índice 0 até o 4
print(arrI[:6]) # slicing: retorna os elementos do índice 0 até o 5
print(arrI[5:]) # slicing: retorna os elementos do índice 5 até o final
```

**View: cuidado real aqui**

Essa parte é muito importante. Em NumPy, um slice pode ser uma **view**, ou seja, uma visão do array original.

Então se eu altero o slice, posso acabar alterando o array original também. Isso é perigoso se eu estiver mexendo em dados de treino ou features originais sem perceber.

```python
print(arrI)

slice_of_arrI = arrI[0:5]
slice_of_arrI[:] = 100

print(slice_of_arrI)
print(arrI)#O slice_of_arrI é uma view do arrI, então quando alteramos o slice_of_arrI, o arrI também é alterado
```

**Copy**

Quando eu quero uma cópia independente, uso `.copy()`.

Essa diferença entre view e copy é importante para não contaminar dados sem querer.

```python
arrI_copy = arrI.copy() #Diferente do slice, o copy cria uma cópia do array, então quando alteramos o copy, o arrI não é alterado
```

```python
arrI_copy[:] = 200
```

```python
print(arrI)
print(arrI_copy)
```

**Moral da história: view vs copy**

Essa parte parece detalhe, mas não é.

Se eu estiver preparando um dataset e alterar uma view achando que é cópia, posso modificar dados originais sem perceber. Isso pode gerar erro difícil de achar depois.

Regra prática: quando for mexer em um pedaço e eu não quiser alterar o original, uso `.copy()`.

**Arrays 2D**

Agora entramos em arrays bidimensionais. Isso é diretamente ligado à ideia de imagem.

Uma imagem em escala de cinza pode ser vista como uma matriz 2D: linha, coluna e valor do pixel.

```python
arr_2d = np.array([[5,10,15],[20,25,30],[35,40,45]])
```

```python
arr_2d
```

**Indexação em matriz**

Aqui aparecem duas formas de acessar valores:

- `arr_2d[linha][coluna]`
- `arr_2d[linha, coluna]`

A segunda forma costuma ser mais comum em NumPy.

```python
print(arr_2d[1][1]) #Retorna o elemento da segunda linha e segunda coluna, que é 25
print(arr_2d[0]) #Retorna a primeira linha do array, que é [5,10,15]
print(arr_2d[0:2]) #Retorna as duas primeiras linhas do array, que são [[5,10,15],[20,25,30]]
print(arr_2d[2][1]) #Retorna o elemento da terceira linha e segunda coluna, que é 40

# Outra notação para acessar elementos de arrays 2D é usando vírgula para separar os índices de linha e coluna:

print(arr_2d[1,1]) #Retorna o elemento da segunda linha e segunda coluna, que é 25
print(arr_2d[:2,1:]) #Retorna as duas primeiras linhas e todas as colunas a partir da segunda coluna, que são [[10,15],[25,30]]
print(arr_2d[1:,:2]) #Retorna todas as linhas a partir da segunda linha e as duas primeiras colunas, que são [[20,25],[35,40]]
```

## NumPy Operations

Nesta parte entram operações matemáticas com arrays.

- Array com Array
- Array com Escalares
- Funções universais do NumPy

A moral aqui é: o NumPy aplica a operação em vários elementos de uma vez. Eu não preciso fazer um `for` para somar, multiplicar, elevar ao quadrado ou tirar raiz de cada elemento. Isso é muito útil quando estou lidando com milhares de features ou muitos bytes de uma amostra.

```python
import numpy as np
```

**Operações entre arrays**

Quando faço `arr + arr`, o NumPy soma elemento com elemento.

Isso é bem diferente de lista comum do Python. Com NumPy, a operação é vetorizada.

```python
arr = np.arange(0,11)
```

```python
arr + arr
```

```python
arr - arr
```

```python
arr * arr
```

**Array com escalar**

Aqui cada elemento do array recebe a mesma operação.

Isso é muito usado para normalização. Por exemplo, bytes de 0 a 255 podem ser divididos por 255 para virar valores entre 0 e 1 antes de entrar em um modelo.

```python
print(arr + 100)
print(arr * 2)
print(arr ** 2) #Eleva cada elemento do array ao quadrado
```

**Normalização**

Essa parte é importante para machine learning.

Se os valores vêm de bytes, normalmente estão entre 0 e 255. Muitos modelos funcionam melhor quando os dados são normalizados, por exemplo entre 0 e 1.

Exemplo conceitual:

`bytes_normalizados = bytes_array / 255`

Isso não está no notebook original, mas é uma ideia muito importante para o projeto.

**Divisão por zero**

Aqui aparece um aviso importante: dividir zero por zero gera `nan`.

`nan` significa “not a number”. Em dataset real, esse tipo de valor pode ferrar treino, métrica e análise se eu não tratar direito.

```python
arr / arr #retorna nan para o elemento 0, pois não é possível dividir por 0
```

**Funções universais**

Funções como `sqrt`, `exp`, `log` e `sin` são aplicadas em todos os elementos do array.

Isso é útil para transformação matemática de features, embora eu precise sempre pensar se a transformação faz sentido para o problema.

```python
np.sqrt(arr) #Retorna a raiz quadrada de cada elemento do array
```

```python
np.exp(arr) #Retorna o exponencial de cada elemento do array
```

```python
np.log(arr) #Retorna o logaritmo natural de cada elemento do array
```

```python
np.sin(arr) #Retorna o seno de cada elemento do array
```
